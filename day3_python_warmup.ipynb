```python
#A LIST OF CUSTOMERS
customers = [
    {"name":"Alice","country":"USA","spent": 150},
    {"name":"Bob","country":"Brazil","spent": 89},
    {"name":"Charlie","country":"USA","spent": 230},
    {"name":"Diana","country":"Germany","spent": 412},
    {"name":"Edward","country":"Brazil","spent": 95},
    {"name":"Fiona","country":"Germany","spent": 178},
    {"name":"George","country":"USA","spent": 65},
    {"name":"Hana","country":"Japan","spent": 320},
]
#SHOW ALL CUSTOMERS (like SELECT * FROM Customer)
for c in customers:
    print (c)
```

```python
#JUST CUSTOMERS FROM USA (like WHERE country = 'USA')
usa_customers = [c for c in customers if c["country"] == "USA"]

print("USA customers:")
for c in usa_customers:
    print(c["name"], "-", c["country"], "-","$" + str(c["spent"]))
```

```python
#CUSTOMERS SORTED BY SPENDING,HIGHEST FIRST(like ORDER BY spent DESC)
sorted_customers = sorted(customers, key=lambda c: c["spent"], reverse=True)

print("Customers ranked by spending:")
for c in sorted_customers:
    print(f"{c['name']:10} ${c['spent']}")
```

```python
#HOW MANY CUSTOMERS DO WE HAVE (like SELECT COUNT(*) FROM customers)
total = len(customers)
print(f"Total customers: {total}")

#HOW MANY ARE FROM USA
usa_count = len([c for c in customers if c["country"] == "USA"])
print(f"USA customers: {usa_count}")

#HOW MANY SPENT OVER $100
big_spenders = len([c for c in customers if c["spent"] > 100])
print(f"Customers spending over $100: {big_spenders}")
```

```python
#TOTAL SPENDING PER COUNTRY (like GROUP BY country, SUM(spent))
country_totals = {}

for c in customers:
    country = c["country"]
    if country not in country_totals:
        country_totals[country] = 0
    country_totals[country] += c["spent"]

print("Total spending by country:")
for country,total in country_totals.items():
    print(f" {country:10} ${total}")
```
